```python

import pandas as pd
from textblob import TextBlob
from colorama import Fore, init
import random

init(autoreset=True)

# Movie database
movies = {
    "Title": [
        "Inception", "Interstellar", "Titanic",
        "The Avengers", "The Notebook",
        "Joker", "Frozen", "The Conjuring"
    ],
    "Genre": [
        "Sci-Fi", "Sci-Fi", "Romance",
        "Action", "Romance", "Drama",
        "Animation", "Horror"
    ],
    "IMDb Rating": [
        8.8, 8.7, 7.9, 8.0, 7.8, 8.3, 7.4, 7.5
    ],
    "Overview": [
        "A skilled thief enters people's dreams to steal secrets.",
        "A team travels through space to find a new home for humanity.",
        "A beautiful love story develops aboard a doomed ship.",
        "Superheroes join forces to save the world.",
        "A romantic story about everlasting love.",
        "A lonely man transforms into a dangerous criminal.",
        "A princess discovers the power of love and friendship.",
        "A family experiences terrifying supernatural events."
    ]
}

# Create DataFrame
df = pd.DataFrame(movies)

# Display movie details
def display_movies(data):

    if data.empty:
        print(Fore.RED + "\nNo movies found!")
        return

    for index, movie in data.iterrows():

        # Sentiment analysis
        polarity = TextBlob(movie["Overview"]).sentiment.polarity

        if polarity > 0:
            sentiment = "Positive"
        elif polarity < 0:
            sentiment = "Negative"
        else:
            sentiment = "Neutral"

        print(Fore.YELLOW + "\n-----------------------------")
        print(Fore.CYAN + "Title:", movie["Title"])
        print("Genre:", movie["Genre"])
        print("IMDb Rating:", movie["IMDb Rating"])
        print("Overview:", movie["Overview"])
        print(Fore.GREEN + "Sentiment:", sentiment)

# Main menu
while True:

    print(Fore.MAGENTA + "\n===== AI MOVIE RECOMMENDER =====")
    print("1. Recommend by Genre")
    print("2. Recommend by Mood")
    print("3. Recommend by IMDb Rating")
    print("4. Random Movie Suggestion")
    print("5. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":

        genre = input("Enter genre (Action, Romance, Sci-Fi, Drama, Horror, Animation): ").strip()

        result = df[df["Genre"].str.lower() == genre.lower()]
        display_movies(result)

    elif choice == "2":

        mood = input("How are you feeling? (Happy/Sad/Excited): ").lower()

        if mood == "happy":
            result = df[df["Overview"].apply(
                lambda x: TextBlob(x).sentiment.polarity > 0
            )]

        elif mood == "sad":
            result = df[df["Overview"].apply(
                lambda x: TextBlob(x).sentiment.polarity < 0
            )]

        elif mood == "excited":
            result = df[df["Genre"].isin(["Action", "Sci-Fi"])]

        else:
            print(Fore.RED + "Invalid mood!")
            continue

        display_movies(result)

    elif choice == "3":

        rating = float(input("Enter minimum IMDb rating (0 to 10): "))

        result = df[df["IMDb Rating"] >= rating]
        display_movies(result)

    elif choice == "4":

        movie = df.sample(1)
        display_movies(movie)

    elif choice == "5":

        print(Fore.GREEN + "\nThank you for using AI Movie Recommender!")
        break

    else:
        print(Fore.RED + "Invalid choice! Please try again.")
```
